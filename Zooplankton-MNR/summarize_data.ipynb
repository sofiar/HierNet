```python
import pandas as pd
from hierclassifier.samples_setup import ImageDataset, merge_classes
```

### Get data - Zooplankton

```python
DataPath = '/data/Hier-Zooplankton'


RESOLUTION = 65
SEED = 6565
```

### Crate ImageDataset


```python
SELECTED_LIST = [
    '149um>',
    'Asplanchna',
    'Bosminidae',
    'Brachionidae',
    'Bubbles',
    'Bythotrephes',
    'Calanoid',
    'Ceriodaphnia',
    'Chydoridae',
    'Chydorus',
    'Cladocera',
    'Conochilus',
    'Copepoda',
    'Cyclopoid',
    'Daphnia',
    'Debris',
    'Diaphonosoma',
    'Duplicates',
    'Eubosmina',
    'Exoskeleton',
    'Fiber_Hairlike',
    'Fiber_Squiggly',
    'Harpacticoid',
    'Holopedium',
    'Kellicottia',
    'Keratella',
    'Leptodora',
    'MultiClass',
    'Nauplius_Copepod',
    'Partial_Copepod',
    'Partial_Daphnia',
    'Partial_Leptodora',
    'Partial_Other',
    'Phytoplankton',
    'Plant_Matter',
    'Ploima',
    'Polyarthra',
    'Rotifer',
    'Sididae',
    'Synchaetidae',
    'Trichocerca',
    'Unknown',
    'Unknown_Small'        
    ]

dataset_selected = ImageDataset(
    data_directory = DataPath,
    class_names = SELECTED_LIST,
    max_class_size = 15000,
    image_resolution = 64,
    image_transforms = None,
    format_file = '.tif',
    seed = SEED
    )
```

### Print details


```python
data_details = dataset_selected.get_dataset_details()
dataset_selected.print_dataset_details()
```

### Plot


```python
import matplotlib.pyplot as plt 
df = pd.DataFrame(data_details)

ax = df.plot(x='Class',y='Counts', kind='bar',legend = False, figsize=(12, 5))
for p in ax.patches:
    ax.annotate(
        str(int(p.get_height())),                
        (p.get_x() + p.get_width() / 2, p.get_height()),  
        ha="center", va="bottom"                
    )

plt.ylabel('Counts')
plt.title('Class Distribution')
plt.show()
```
